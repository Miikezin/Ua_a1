##lista de convidados

```python
convidados = ("alice","rodrigo","carol","felipe")
confirmados = ["alice","rodrigo"]
```

```python
print("convidados que ainda não confirmaram:")
for convidado in convidados:
    if convidado not in confirmados:
        print(f"-{convidado}")
```

##U2_a2 conjuntos

```python
meu_conjunto = set()
```

```python
meu_conjunto.add(10)
```

```python
meu_conjunto.add(20)
meu_conjunto.add(30)
```

```python
meu_conjunto.remove(20)
```

```python
meu_conjunto
```

```python
elemento = 20
if elemento in meu_conjunto:
  print(f"o elemento{elemento} está no conjunto")
else:
  print(f"o elemento{elemento} não esta no conjunto")
```
