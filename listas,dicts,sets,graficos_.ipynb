```python
import matplotlib.pyplot as plt
```

```python
class produto:
  def __init__(self,nome,preco,categoria,estoque):
      self.nome = nome
      self.preco = preco
      self.categoria = categoria
      self.estoque = estoque

  #definição de como o produto aparece quando é impresso

  def __str__(self):
    return (f"{self.nome} | R$ {self.preco:.2f} | Categoria: {self.categoria} | Estoque: {self.estoque}")
```

```python
catalogo = []
categorias = []
```

```python
def adicionar_produto(nome,preco,categoria,estoque):
  produto_novo = produto(nome,preco,categoria,estoque)
  catalogo.append(produto_novo)
  categorias.append(categoria)
  print(f"produto '{nome}' adicionado ao catalogo.")
```

```python
def listar_catalogo():
  print("\n---catalogo da Minha loja---")
  for produto in catalogo:
    print(produto)
```

```python
#cadastrar produtos

catalogo = []
categorias = []

adicionar_produto("notebook",  3500.00,"informatica",   12)
adicionar_produto("mouse",      25.00,"informatica",   15)
adicionar_produto("smartphone",    2200.00,"informatica",  35)
adicionar_produto("cafeteira",    350.00,"eletrodomesticos",   10)
adicionar_produto("fone bluetooth",  80.00,"informatica",  40)
adicionar_produto("geladeira", 3500.00,"eletrodomesticos",   15)

listar_catalogo()
```

```python
#grafico por categoria
categorias_unicas = list(set(categorias))
categorias_unicas.sort()

#contagem de produtos em cada categoria(list comprehension)
contagem = [categorias.count(cat)for cat in categorias_unicas]
```

```python
plt.bar(categorias_unicas,contagem,color='royalblue')
plt.xlabel('categoria')
plt.ylabel('numero de produtos')
plt.title('Minha loja-produtos por Categoria')
```
