##lista de convidados

```python
convidados = ("alice","rodrigo","carol","felipe")
confirmados = ["alice","rodrigo"]
```

```python
print("convidados que ainda não confirmaram:")
for convidado in convidados:
    if convidado not in confirmados:
        print(f"-{convidado}")
```

##bibliotecas e modulos

```python
#primeira forma de import
import math

print(math.sqrt(25))  #raiz quadrada
print(math.log2(1024)) #logaritimo na base 2
print(math.cos(45)) #cosseno
```

```python
#segunda forma -import math as m
import math as s

print(s.sqrt(25))
print(s.log2(1024))
print(s.cos(45))
```

```python
#terceira forma-from math import
from math import sqrt,log2,cos

print(sqrt(25))
print(log2(1024))
print(cos(45))
```

##biblioteca matplotlib

```python
#Matplotlib-grafico de linha
import matplotlib.pyplot as plt
#dados
x = [1,2,3,4,5]
y = [2,4,1,3,5]

#criar um gráfico de linha
plt.plot(x,y)

#adicionar rótulos aos eixos
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

#adicionar um título
plt.title('grafico de linha')

#exibir
plt.show()
```
